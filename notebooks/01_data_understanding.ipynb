# Data Understanding & Quality Audit

## Objective

Conducted an initial data-quality and structural audit of the raw retail transaction dataset before performing any transformation or analysis. The objective was to understand the dataset, identify data-quality risks, validate key business relationships, and establish a reliable foundation for downstream analysis and dashboard development.

## Dataset Profile

- **Records:** 50,150
- **Attributes:** 19
- **Domain:** Retail Sales
- **Data Coverage:** Orders, customers, products, geography, pricing, discounts, sales, costs, profitability, payment methods, sales channels, and order status.

## Audit Performed

### Data Structure & Types
- Reviewed dataset dimensions, column structure, data types, and cardinality.
- Validated that core numerical measures such as Quantity, Unit_Price, Discount, Sales, Cost, and Profit are stored using appropriate numerical datatypes.
- Identified `Order_Date` as a date field requiring standardization because it is currently stored as text and contains multiple date representations.

### Data Completeness
- Profiled missing values and their distribution across columns.
- Missing data was identified in fields including `Customer_Name`, `City`, `Discount`, and `Payment_Method`.

### Data Integrity
- Identified duplicate transaction records and verified duplicate rows at the record level.
- Investigated numerical ranges and identified potentially invalid values in:
  - Quantity
  - Unit_Price
  - Discount
- Extreme values were flagged for further investigation rather than removed without justification.

### Categorical Consistency
- Audited categorical fields including Region, Category, Payment_Method, Sales_Channel, and Order_Status.
- Identified inconsistent text representations, including capitalization/formatting variations that could cause incorrect grouping and aggregation during analysis.

### Date Consistency
- Identified multiple date representations within `Order_Date`, including records containing timestamps and records containing date-only values.
- This inconsistency could affect time-based analysis and therefore requires standardization.

### Business Logic Validation
Key financial relationships were tested against sample records:

**Sales = Quantity × Unit_Price × (1 − Discount)**

**Profit = Sales − Cost**

The calculated profit values matched the recorded profit values for the sample tested, providing an initial validation of the dataset's financial calculations.

An additional business-rule observation was identified: **Cancelled and Returned orders contain positive Sales values**. This was documented for further investigation rather than being automatically classified as an error.

## Key Outcome

The audit established that the raw dataset contains several data-quality challenges that could affect analytical accuracy, including missing values, duplicate records, invalid numerical values, inconsistent categorical representations, mixed date formats, and business-rule exceptions.

The raw source file was preserved without modification, and all identified issues were documented for systematic cleaning and validation in the subsequent data-preparation stage.
