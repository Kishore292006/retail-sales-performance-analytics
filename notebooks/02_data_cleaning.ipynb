# Data Cleaning & Preparation

## Objective

Transform the audited raw retail transaction data into a **consistent, reliable, and analysis-ready dataset** while preserving the original source data.

The cleaning process addresses data-quality issues identified during the initial audit, including missing values, duplicate records, inconsistent categorical values, mixed date formats, invalid numerical values, and business-rule anomalies.

## Data Preparation Approach

The cleaning process follows a controlled and reproducible workflow:

- Preserve the raw dataset as the source of truth.
- Perform transformations on a separate working copy.
- Apply business-context-driven cleaning decisions rather than indiscriminate row deletion.
- Validate data after each major transformation.
- Document significant assumptions and treatment decisions.
- Maintain a reproducible path from raw data to the final cleaned dataset.

## Cleaning Scope

1. Duplicate detection and removal
2. Categorical value standardization
3. Date parsing and standardization
4. Missing-value treatment
5. Numerical data validation
6. Business-rule validation
7. Final data-quality verification

## Deliverable

A validated, analysis-ready dataset suitable for:

- Exploratory Data Analysis
- Business analysis and KPI development
- Power BI reporting
- Reusable Python data-processing workflows

# Final Data Cleaning Summary

## Cleaning Outcome

The raw retail transaction dataset was cleaned and transformed into an analysis-ready dataset while preserving the original source data.

### Final Dataset

- **Rows:** 49,975
- **Columns:** 19
- **Missing values:** 0
- **Duplicate records:** 0

### Cleaning Performed

- Preserved the original raw dataset and created a separate working copy.
- Standardized all column names to lowercase.
- Removed 150 exact duplicate records.
- Standardized inconsistent categorical values across Region, Payment Method, and Sales Channel.
- Converted mixed Order Date representations into a standardized datetime format.
- Treated missing Customer Name, City, and Payment Method values as `Unknown`.
- Treated missing Discount values as `0`.
- Removed records containing invalid quantities, unit prices, or discount values.
- Validated the relationships between Quantity, Unit Price, Discount, Sales, Cost, and Profit.
- Investigated financial inconsistencies between calculated and recorded Sales and Profit values without silently overwriting the original financial fields.
- Validated business rules related to order status, profitability, cost, quantity, price, and discount.
- Performed final data-quality validation after all cleaning operations.
- Saved the cleaned dataset as `data/processed/sales_cleaned.csv`.
